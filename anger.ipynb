# Data Cleaning and Preprocessing

```python
# Data Cleaning and Preprocessing
f.find_nan(anger,'Ways_to_Express')
anger['Ways_to_Express'].fillna('Not applicable', inplace=True)
anger
```

```python
# Encoding
categories= [['No, not really','Yes, completely','Sometimes','Partly','Rarely']]
encoder= pp.OrdinalEncoder(categories=categories, dtype=int)
# Map the text responses to ordered numeric values before casting
meaning_map = {
    'No, not really': 0,
    'Yes, completely': 1,
    'Sometimes': 2,
    'Partly': 3,
    'Rarely': 4
}

anger['Meaning'] = anger['Meaning'].apply(
    lambda x: meaning_map.get(str(x).strip(), x)
).astype(int)

anger['Meaning'] = encoded.astype(int)
anger
```

```python
anger['Better_Way'].fillna('Not applicable', inplace=True)
anger['Trigger'].fillna('Feeling disrespected', inplace= True)
anger
```

```python
category= [['Never','Rarely','Sometimes','Often','Always']]
enc=pp.OrdinalEncoder(categories=category, dtype=int)
encd= enc.fit_transform(anger[['Reflection']])

anger['Reflection'] = encd.astype(int)
anger
```

```python
# Creating function for encoding
def encode(df, column_name, categories):
    """
    Encodes a categorical column in the DataFrame using Ordinal Encoding.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the column to encode.
    column_name (str): The name of the column to encode.
    categories (list): A list of lists specifying the order of categories.

    Returns:
    pd.DataFrame: The DataFrame with the encoded column.
    """
    encoder = pp.OrdinalEncoder(categories=categories, dtype=int)
    encoded = encoder.fit_transform(df[[column_name]])
    df[column_name] = encoded.astype(int)
    return df
```
