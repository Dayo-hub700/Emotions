# Data Dictionary

- `When you ecome angry during a conversation, how do you usually react?` - `Reaction`
  
- `What type of conversations or situations are most likely to trigger your anger?` - `Trigger`
  
- `How do you usually handle your anger after the conversation?` - `Handling`
  
- `When you say things while angry, do you usually mean them?` - `Meaning`
  
- `If you answered “Yes,” “Partly,” or “Sometimes,” do you think there could have been a better or calmer way to express what you meant?` - `Better_Way`
  
- `If you answered “Rarely” or “No,” do you think your anger caused you to say things you did not truly intend to express? ` - `Ways_to_Express`
  
- `After calming down, how often do you feel your words came out harsher than intended? ` - `Reflection`

```python
# Encoding
categories= [['No, not really','Yes, completely','Sometimes','Partly']]
encoder= pp.OrdinalEncoder(categories=categories, dtype=int)
encoded= encoder.fit_transform(anger[['Meaning']])

anger['Meaning'] = encoded.astype(int)
```
