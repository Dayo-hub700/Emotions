# Data Cleaning and Preprocessing

```python
# Data Cleaning and Preprocessing
f.find_nan(anger,'Ways_to_Express')

#anger
```

```python
# Encoding
categories= [['No, not really','Yes, completely','Sometimes','Partly','Rarely']]
encoder= pp.OrdinalEncoder(categories=categories, dtype=int)
encoded= encoder.fit_transform(anger[['Meaning']])

anger['Meaning'] = encoded.astype(int)
anger
```

```python
anger['Better_Way'].fillna('Not applicable', inplace=True)
anger['Trigger'].fillna('Feeling disrespected', inplace= True)
anger
```
