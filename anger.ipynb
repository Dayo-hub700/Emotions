# Data Cleaning and Preprocessing

```python
# Data Cleaning and Preprocessing
f.find_nan(anger,'Ways_to_Express')

#anger
```

```python
# Encoding
categories= [['No, not really','Yes, completely','Sometimes','Partly','Rarely']]
encoder= pp.OrdinalEncoder(categories=categories, dtype=int)
encoded= encoder.fit_transform(anger[['Meaning']])

anger['Meaning'] = encoded.astype(int)
anger
```

```python
anger['Better_Way'].fillna('Not applicable', inplace=True)
anger['Trigger'].fillna('Feeling disrespected', inplace= True)
anger['Ways_to_Express'].fillna('Not applicable', inplace=True)
anger
```

```python
anger.drop(columns='Anger_Changes',inplace=True)
anger.head()
```

```python
anger['Trigger'] = anger['Trigger'].replace(
    'Lies , dishonesty, trying to act smart,criticism, betrayal ',
    'Lies, criticism, betrayal',
    regex=False
).str.strip()

anger['Trigger']= anger['Trigger'].replace(
    "Can't i pick two frustration, criticism, lies and being misunderstood",
    'Frustration, criticism, lies, being misunderstood',
    regex=False
).str.strip()

anger['Trigger']= anger['Trigger'].replace(
    "When i’m accused of something i didn’t do",
    'Accusations',
    regex=False
).str.strip()


anger['Handling'] = anger['Handling'].replace('More than one option acc  option. 2, 3, 4 and depending 6', 'Omoo', regex=False).str.strip()
anger['Handling']= anger['Handling'].replace('Sometimes I cry too, anger is a painful emotion', 'I cry',regex= False).str.strip()
anger['Handling']= anger['Handling'].replace(
    'I reflect on what I did if I was wrong I apologise and also tell them Thier faults during conversations most time both sides have a truth in whatever they say so we Check ans explain calmly or loudly but if loudly we laugh over it',
    'I reflect on my actions, apologise when wrong, and calmly or humorously discuss both sides faults, ' \
    'recognising that we may each have some truth.', regex= False
).str.strip()
```

```python
anger
```

```python
def encode(df, column_name, categories):
    encoder = pp.OrdinalEncoder(categories=[categories], dtype=int)
    encoded = encoder.fit_transform(df[[column_name]])
    df[column_name] = encoded.astype(int)
    return df

encode(anger, 'Reflection', ['Never', 'Rarely', 'Sometimes', 'Often', 'Always'])

anger['Reaction'] = anger['Reaction'].replace(
    'Be quiet and let them know Thier faults we reflect together after they are done talking then I tell them not to interrupt me when talking',
    'I stay quiet while they speak, reflect together afterward, address their faults, and ask them not to interrupt me when I speak.',
    regex=False
).str.strip()

encode(anger, 'Reaction', [
    'I become quiet and withdraw',
    'It depends',
    'I say things impulsively',
    'I avoid the conversation entirely',
    'I raise my voice',
    'I cry',
    'I speak more emotionally/intensely',
    'I try to explain my feelings calmly',
    'I stay quiet while they speak, reflect together afterward, address their faults, and ask them not to interrupt me when I speak.',
    "It depends on the person if he/she is raising their voice i will but if they aren't i wont"
])

encode(anger, 'Trigger', ['Feeling disrespected',
                          'Being misunderstood',
                          'Repetition/frustration',
                          'Dishonesty or lies',
                          'Feeling ignored',
                          'Criticism',
                          'Accusations',
                          'Feeling disrespected and being gaslit',
                          'Lies, criticism, betrayal',
                          'Disrespecting my parents',
                          'All except lies'
                          'Frustration, criticism, lies, being misunderstood',
                          'Being misunderstood and frustration'
                          ])

encode(anger, 'Handling', [
    'I apologise if necessary',
    'I reflect on what was said',
    'I avoid the person for some time',
    'I regret some of my words',
    'I calm down quickly',
    'Omoo',
    'I leave the scene to think things through',
    'I reflect on my actions, apologise when wrong, and calmly or humorously discuss both sides faults, recognising that we may each have some truth.'
    'I cry'
])
```
